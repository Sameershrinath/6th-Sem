```python
import pandas as pd
import numpy as np

# Sample dataset
data = {
    'Age': [20, 22, 22, np.nan, 25, 100, 30, 28],
    'Salary': [20000, 25000, 25000, 27000, np.nan, 500000, 30000, 28000],
    'City': ['A', 'B', 'B', 'C', 'A', 'A', 'C', 'B']
}

df = pd.DataFrame(data)

# Handle missing values
df.fillna(df.mean(numeric_only=True), inplace=True)

# Remove duplicates
df.drop_duplicates(inplace=True)

# Remove outliers (Age < 60)
df = df[df['Age'] < 60]

# Normalization
df['Salary'] = (df['Salary'] - df['Salary'].min()) / (df['Salary'].max() - df['Salary'].min())

# Encoding
df = pd.get_dummies(df, columns=['City'])

print(df)
```
