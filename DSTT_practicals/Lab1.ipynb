```python
import numpy as np
from scipy import stats

# Sample dataset
data = [10, 12, 15, 18, 20, 22, 25, 30]

# Mean
mean = np.mean(data)

# Median
median = np.median(data)

# Variance
variance = np.var(data)

# Skewness
skewness = stats.skew(data)

# Percentile (25th, 50th, 75th)
p25 = np.percentile(data, 25)
p50 = np.percentile(data, 50)
p75 = np.percentile(data, 75)

# Probability (example: P(value > 20))
prob = len([x for x in data if x > 20]) / len(data)

print("Mean:", mean)
print("Median:", median)
print("Variance:", variance)
print("Skewness:", skewness)
print("Percentiles:", p25, p50, p75)
print("Probability (value > 20):", prob)
```
