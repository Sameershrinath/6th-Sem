# Experiment 3: Stemming and Lemmatization Comparison

## Aim
Perform the experiment using Python in Jupyter Notebook to compare stemming and lemmatization techniques in NLP.

## Theory
Stemming and lemmatization are text normalization techniques used in Natural Language Processing.
Stemming reduces words to their root form by removing suffixes, often using simple rules.
Because it is rule-based and fast, stemming may produce non-dictionary words such as "studies" -> "studi".
Lemmatization converts a word to its meaningful base form (lemma) using vocabulary and morphological analysis.
Lemmatization is usually more accurate than stemming but can be slower due to linguistic processing.
These techniques help reduce vocabulary size, improve matching, and support tasks like search, classification, and sentiment analysis.

## Algorithm / Steps
1. Import required NLTK modules for tokenization, stemming, and lemmatization.
2. Download required NLTK resources (`punkt`, `wordnet`, and `omw-1.4`) if not already available.
3. Define a sample paragraph containing words in different grammatical forms.
4. Perform sentence tokenization and word tokenization.
5. Apply Porter Stemmer to each word token.
6. Apply WordNet Lemmatizer to each word token.
7. Display original tokens, stemmed tokens, and lemmatized tokens for comparison.
8. Observe the quality difference between stemming and lemmatization outputs.

```python
# Import required libraries
import re
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Download NLTK resources (safe to run multiple times)
nltk.download('punkt', quiet=True)
nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)

# Sample text for the experiment
text = (
    "The striped bats are hanging on their feet for best. "
    "The students were studying, studies are important, and studied lessons improve knowledge."
)

# Sentence tokenization
sentences = sent_tokenize(text)

# Word tokenization
word_tokens = word_tokenize(text)

# Keep only alphabetic words and convert to lowercase for clean comparison
clean_tokens = [token.lower() for token in word_tokens if token.isalpha()]

# Initialize stemmer and lemmatizer
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

# Apply stemming and lemmatization
stemmed_words = [stemmer.stem(word) for word in clean_tokens]
lemmatized_words = [lemmatizer.lemmatize(word) for word in clean_tokens]

# Display outputs
print("Original Text:\n", text)
print("\nSentence Tokens:")
for i, sent in enumerate(sentences, start=1):
    print(f"{i}. {sent}")

print("\nClean Word Tokens:")
print(clean_tokens)

print("\nStemmed Words:")
print(stemmed_words)

print("\nLemmatized Words:")
print(lemmatized_words)

print("\nComparison (Word | Stem | Lemma):")
for w, s, l in zip(clean_tokens, stemmed_words, lemmatized_words):
    print(f"{w:12} | {s:12} | {l}")
```

## Conclusion
The experiment shows that both stemming and lemmatization reduce words to base forms, but with different quality.

- **Stemming** (Porter Stemmer) is faster and rule-based, but often produces truncated/non-dictionary forms such as `studi`, `improv`, and `knowledg`.
- **Lemmatization** (WordNet Lemmatizer) gives more meaningful dictionary forms such as `study`, `lesson`, `bat`, and `foot`.

Hence, stemming is useful for quick normalization, while lemmatization is preferred when linguistic accuracy and interpretability are important.
