# Experiment 5: Named Entity Recognition (NER) using spaCy

## Aim
Perform Named Entity Recognition (NER) on sample text using the spaCy library in Python.

## Theory
Named Entity Recognition (NER) is a subtask of information extraction that identifies and classifies named entities in text into predefined categories such as person names, organizations, locations, dates, and monetary values.
NER is a key component in NLP pipelines used for tasks like question answering, document summarization, and knowledge graph construction.
spaCy is an open-source NLP library written in Python that provides fast, industrial-strength NER through pre-trained statistical models.
These models are trained on large annotated corpora and can recognize entities with high accuracy using deep learning architectures.
The `en_core_web_sm` model in spaCy recognizes over 18 entity types including PERSON, ORG, GPE (Geo-Political Entity), DATE, MONEY, and more.
NER is distinct from keyword extraction as it identifies the semantic category of words, not just their frequency.
Applications of NER include news analysis, legal document parsing, medical record processing, and customer feedback analysis.

## Algorithm / Steps
1. Install and import the spaCy library and ensure the English model (`en_core_web_sm`) is available.
2. Load the pre-trained `en_core_web_sm` language model into a spaCy `nlp` pipeline object.
3. Define a sample paragraph containing various named entities (persons, places, organizations, dates, etc.).
4. Process the text by passing it through the `nlp` pipeline to create a `doc` object.
5. Iterate over `doc.ents` to extract each recognized named entity.
6. Display each entity's text, its label (entity type), and a human-readable explanation of the label.
7. Visualize the entities inline in the notebook using `spacy.displacy.render` with `style="ent"`.

```python
# ── Step 1: Ensure spaCy and its English model are available ──────────────────
import subprocess, sys

# Install spaCy if not already installed
subprocess.run([sys.executable, "-m", "pip", "install", "spacy", "--quiet"], check=True)

# Download the small English model if not already present
subprocess.run([sys.executable, "-m", "spacy", "download", "en_core_web_sm", "--quiet"], check=True)

# ── Step 2: Import spaCy and load the pre-trained English model ───────────────
import spacy
from spacy import displacy

nlp = spacy.load("en_core_web_sm")

# ── Step 3: Define a sample paragraph rich in named entities ──────────────────
text = (
    "Apple Inc. was founded by Steve Jobs and Steve Wozniak in Cupertino, California in 1976. "
    "The company is headquartered in the United States and has offices in London, Tokyo, and Sydney. "
    "In 2023, Apple reported a revenue of $394.33 billion. "
    "Elon Musk, CEO of Tesla and SpaceX, visited the NASA headquarters in Washington D.C. last Monday."
)

# ── Step 4: Process the text through the NLP pipeline ────────────────────────
doc = nlp(text)

# ── Step 5 & 6: Extract and display named entities ───────────────────────────
print("Input Text:\n", text)
print("\nNamed Entities Detected:")
print(f"{'Entity':<30} {'Label':<12} {'Description'}")
print("-" * 65)
for ent in doc.ents:
    print(f"{ent.text:<30} {ent.label_:<12} {spacy.explain(ent.label_)}")

# ── Step 7: Visualise entities inline in the notebook ────────────────────────
print("\nEntity Visualisation (HTML rendered below):")
displacy.render(doc, style="ent", jupyter=True)
```

## Conclusion
In this experiment, Named Entity Recognition was successfully performed using spaCy's `en_core_web_sm` model.
The model correctly identified and categorised entities such as persons (PERSON), organisations (ORG), locations (GPE), dates (DATE), and monetary values (MONEY) from the input text.
The `displacy` visualiser provided a clear, colour-coded HTML view of entities in-context, making results easy to interpret.
This demonstrates that spaCy's pre-trained NER pipeline is an accurate and efficient tool for structured information extraction from unstructured text.
