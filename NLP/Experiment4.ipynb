# Experiment 4: Part of Speech (POS) Tagging

## Aim
Perform Part of Speech (POS) tagging using Python in Jupyter Notebook.

## Theory
Part of Speech (POS) tagging is the process of assigning grammatical categories to words in a sentence.
Common POS tags include noun (NN), verb (VB), adjective (JJ), and adverb (RB).
POS tagging is an important NLP step because it helps machines understand sentence structure.
It is used in applications such as information extraction, parsing, question answering, and text analytics.
NLTK provides the `pos_tag()` function to automatically assign POS tags to tokenized words.
The quality of POS tagging depends on both tokenization and the trained tagging model.
Accurate POS tags improve many downstream NLP tasks by adding syntactic information.

## Algorithm / Steps
1. Import NLTK and required tokenization/tagging functions.
2. Define a sample sentence for POS tagging.
3. Download required NLTK resources if they are not already available.
4. Tokenize the sentence into individual words.
5. Apply POS tagging on the tokenized words.
6. Display each token with its corresponding POS tag.
7. Analyze the output and verify grammatical labels.

```python
# Import NLTK and required functions
import nltk
from nltk.tokenize import word_tokenize
from nltk import pos_tag

# Define sample text
text = "The quick brown fox jumps over the lazy dog near the river bank."

# Download required NLTK resources (safe if already installed)
nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)
nltk.download("averaged_perceptron_tagger", quiet=True)
nltk.download("averaged_perceptron_tagger_eng", quiet=True)

# Tokenize the sentence into words
tokens = word_tokenize(text)

# Apply POS tagging on tokens
tagged_tokens = pos_tag(tokens)

# Display input text
print("Input Text:\n", text)

# Display token list
print("\nWord Tokens:")
print(tokens)

# Display POS tagging result in a readable format
print("\nPOS Tags (Word -> Tag):")
for word, tag in tagged_tokens:
    print(f"{word:>10} -> {tag}")
```

## Conclusion
In this experiment, POS tagging was successfully performed using NLTK.
The sentence was tokenized, and each word was assigned a grammatical tag.
The output demonstrates how NLP systems identify syntactic roles of words in context.
POS tagging is a key preprocessing step for advanced tasks like parsing, information extraction, and machine translation.
