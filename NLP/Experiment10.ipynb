# Experiment 10: Simple Rule-Based NLP Chatbot

## Aim
To implement a simple rule-based chatbot using Python in Jupyter Notebook.

## Theory
A rule-based chatbot is a basic conversational system that responds to user input based on predefined rules and patterns. It does not use machine learning but relies on keyword matching and conditional statements. When a user enters a query, the chatbot searches for specific keywords and returns a corresponding response. These chatbots are simple, fast, and easy to implement but lack the ability to understand context or learn from interactions. They are commonly used in FAQs, customer support, and basic automation tasks. Despite their limitations, they are useful for understanding the fundamentals of conversational AI.

## Algorithm / Steps

1. Define a set of predefined rules (keywords and responses).
2. Take user input from the console.
3. Convert input to lowercase for uniform processing.
4. Check for keywords in the input.
5. Match the input with predefined rules.
6. Return the corresponding response.
7. Repeat until the user exits the chatbot.

```python
# Simple Rule-Based Chatbot

def chatbot_response(user_input):
    user_input = user_input.lower()
    
    # Rule-based responses
    if "hello" in user_input or "hi" in user_input:
        return "Hello! How can I help you?"
    
    elif "how are you" in user_input:
        return "I'm just a chatbot, but I'm doing great!"
    
    elif "your name" in user_input:
        return "I am a simple NLP chatbot."
    
    elif "help" in user_input:
        return "I can answer basic questions. Try asking something!"
    
    elif "bye" in user_input or "exit" in user_input:
        return "Goodbye! Have a nice day."
    
    else:
        return "Sorry, I didn't understand that."
```

```python
# Run chatbot

print("Chatbot is running! Type 'exit' to stop.\n")

while True:
    user_input = input("You: ")
    
    response = chatbot_response(user_input)
    print("Bot:", response)
    
    if "exit" in user_input.lower():
        break
```

## Output

- The chatbot takes user input from the console.
- It matches keywords with predefined rules.
- Based on the input, it generates appropriate responses.
- The chatbot continues conversation until the user types 'exit'.

## Conclusion

The rule-based chatbot successfully responds to user queries using predefined rules. It demonstrates the basic working of conversational systems without using machine learning. While it is simple and efficient, it cannot handle complex queries or understand context. This experiment helps in understanding the foundation of chatbot development. More advanced chatbots use machine learning and NLP techniques for better interaction.
