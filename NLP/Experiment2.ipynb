# Experiment 2: Word and Sentence Tokenization using NLTK

## Aim
Perform word and sentence tokenization using Python in Jupyter Notebook with the NLTK library.

## Theory
Tokenization is the process of breaking text into smaller meaningful units such as sentences and words.
In Natural Language Processing (NLP), sentence tokenization helps identify sentence boundaries in a paragraph.
Word tokenization splits each sentence into individual terms, punctuation symbols, or tokens.
NLTK provides built-in tokenizers such as `sent_tokenize` and `word_tokenize` for accurate text segmentation.
Sentence-level tokens are useful for tasks like summarization and sentiment analysis at sentence granularity.
Word-level tokens are used in frequency analysis, text classification, and language modeling.
Correct tokenization improves downstream NLP performance because models receive cleaner and more structured input.

## Algorithm / Steps
1. Import required NLTK tokenization functions.
2. Define a sample paragraph containing multiple sentences.
3. Download tokenizer resources if not already available.
4. Apply sentence tokenization to split the paragraph into sentences.
5. Apply word tokenization to split the paragraph into word-level tokens.
6. Print sentence tokens and word tokens.
7. Observe and verify that each sentence and word is segmented correctly.

```python
# Import sentence and word tokenization utilities from NLTK
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize

# Define a sample paragraph with multiple sentences
text = (
    "Natural Language Processing is an exciting field. "
    "It helps computers understand human language. "
    "Tokenization is one of the first preprocessing steps in NLP!"
)

# Download required tokenization resources (safe if already present)
nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)

# Perform sentence tokenization
sentence_tokens = sent_tokenize(text)

# Perform word tokenization
word_tokens = word_tokenize(text)

# Display original text and tokenization outputs
print("Original Text:\n", text)
print("\nSentence Tokens:")
for i, sentence in enumerate(sentence_tokens, start=1):
    print(f"{i}. {sentence}")

print("\nWord Tokens:")
print(word_tokens)

print("\nTotal Sentences:", len(sentence_tokens))
print("Total Word Tokens:", len(word_tokens))
```

## Conclusion
In this experiment, sentence and word tokenization were successfully performed using NLTK.
The text was split into meaningful sentence units and smaller word-level tokens.
This confirms that tokenization is a foundational preprocessing step for NLP tasks.
The generated tokens can now be used for further analysis such as stopword removal, stemming, and feature extraction.
