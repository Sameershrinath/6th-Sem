# Experiment 6: Bag of Words (BoW) Model Implementation

## Aim
To implement the Bag of Words (BoW) model using Python in Jupyter Notebook.

## Theory
The Bag of Words (BoW) model is a fundamental technique in Natural Language Processing (NLP) used to convert text data into numerical form. It represents text as a collection of words without considering grammar or word order. Each unique word in the dataset is treated as a feature, and the frequency of each word is counted. This results in a vector representation for each document. BoW is simple and efficient but ignores context and semantics. It is widely used in text classification, sentiment analysis, and information retrieval tasks. Despite its limitations, it serves as a strong baseline model in NLP.

## Algorithm / Steps

1. Collect the dataset (documents).
2. Convert text into lowercase.
3. Tokenize the text into words.
4. Create a vocabulary of unique words.
5. Count frequency of each word in each document.
6. Represent each document as a vector.
7. Display the Bag of Words matrix.

```python
# Import required library
from sklearn.feature_extraction.text import CountVectorizer

# Dataset
documents = [
    "NLP is very interesting",
    "Machine learning is powerful",
    "NLP and machine learning are related",
    "Deep learning is part of AI",
    "AI is transforming the world",
    "Students are learning NLP",
    "Python is used for machine learning",
    "NLP applications are growing",
    "Data science uses machine learning",
    "AI and NLP are closely related"
]

# Initialize CountVectorizer
vectorizer = CountVectorizer()

# Fit and transform data
X = vectorizer.fit_transform(documents)

# Get vocabulary
features = vectorizer.get_feature_names_out()

# Convert into array
bow_matrix = X.toarray()

# Print results
print("Vocabulary:\n", features)
print("\nBag of Words Matrix:\n", bow_matrix)
```

## Output

- Vocabulary: List of all unique words from the dataset.
- Bag of Words Matrix: Numerical representation of documents.
- Each row represents a document.
- Each column represents a word from the vocabulary.
- Values indicate the frequency of words in each document.
