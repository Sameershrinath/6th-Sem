# Experiment 1: Text Preprocessing

**Aim:** Perform text preprocessing using Python with lowercasing, tokenization, and stopword removal.

## Theory

Text preprocessing is an important step in Natural Language Processing (NLP).
Raw text often contains inconsistent casing, punctuation, and common words that add little meaning.
Lowercasing converts all words to a common format, improving consistency.
Tokenization splits text into smaller units (tokens), usually words.
Stopword removal eliminates frequent words like "the", "is", and "and" that may not help analysis.
These steps reduce noise and prepare text for tasks such as classification and sentiment analysis.
A clean and normalized text representation helps models perform better.

## Algorithm / Steps

1. Take an input sentence or paragraph.
2. Convert the text to lowercase.
3. Tokenize the text into words.
4. Define a stopword list.
5. Remove stopwords from the token list.
6. Display intermediate and final outputs.

```python
# Experiment 1: Text Preprocessing
# Steps covered: lowercasing, tokenization, and stopword removal

import re

# Sample input text
text = "Natural Language Processing is an important area of Artificial Intelligence, and it is very useful in real-world applications."

# Step 1: Lowercasing
lower_text = text.lower()

# Step 2: Tokenization (extract only word tokens)
tokens = re.findall(r"\b[a-z]+\b", lower_text)

# Step 3: Stopword list (basic custom list)
stopwords = {
    "a", "an", "the", "is", "are", "am", "of", "in", "on", "at", "to", "for", "and", "or", "it", "very"
}

# Step 4: Stopword removal
filtered_tokens = [word for word in tokens if word not in stopwords]

# Display results
print("Original Text:")
print(text)
print("\nLowercased Text:")
print(lower_text)
print("\nTokens:")
print(tokens)
print("\nTokens After Stopword Removal:")
print(filtered_tokens)
```

## Conclusion

In this experiment, text preprocessing was performed using lowercasing, tokenization, and stopword removal.
The text became cleaner and more suitable for NLP tasks after removing noise words.
This preprocessing pipeline is a foundational step for building accurate NLP models.
