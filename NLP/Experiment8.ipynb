# Experiment 8: Sentiment Analysis using TextBlob

## Aim
To perform sentiment analysis on text data using the TextBlob library in Python.

## Theory
Sentiment Analysis is a Natural Language Processing (NLP) technique used to determine the emotional tone of a text. It classifies text into categories such as positive, negative, or neutral. TextBlob is a Python library that provides a simple API for common NLP tasks, including sentiment analysis. It uses a rule-based approach along with a predefined lexicon to calculate sentiment polarity and subjectivity. Polarity ranges from -1 (negative) to +1 (positive), while subjectivity ranges from 0 (objective) to 1 (subjective). Sentiment analysis is widely used in social media monitoring, customer feedback analysis, and product reviews.

## Algorithm / Steps

1. Create or collect a dataset of text sentences.
2. Import the TextBlob library.
3. Process each sentence using TextBlob.
4. Calculate sentiment polarity and subjectivity.
5. Classify sentiment as positive, negative, or neutral.
6. Display the results.

```python
from textblob import TextBlob

# Sample dataset
sentences = [
    "I love this product, it is amazing!",
    "This is the worst experience I have ever had.",
    "The movie was okay, not great but not bad.",
    "I am very happy with the service.",
    "The food was terrible and disappointing.",
    "It is just an average performance.",
    "Absolutely fantastic work!",
    "I am not satisfied with the quality.",
    "The experience was wonderful and exciting.",
    "Nothing special, it was normal."
]

# Perform sentiment analysis
for sentence in sentences:
    blob = TextBlob(sentence)
    polarity = blob.sentiment.polarity
    subjectivity = blob.sentiment.subjectivity
    
    # Classify sentiment
    if polarity > 0:
        sentiment = "Positive"
    elif polarity < 0:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"
    
    # Print results
    print("Sentence:", sentence)
    print("Polarity:", polarity)
    print("Subjectivity:", subjectivity)
    print("Sentiment:", sentiment)
    print("-" * 50)
```

## Output

- Each sentence is analyzed for sentiment.
- Polarity:
  - Positive (> 0)
  - Negative (< 0)
  - Neutral (= 0)
- Subjectivity:
  - Measures how subjective or objective the sentence is.
- The model classifies each sentence into Positive, Negative, or Neutral.

## Conclusion

Sentiment analysis using TextBlob is simple and effective for analyzing the emotional tone of text data. It provides polarity and subjectivity scores that help in understanding opinions. The model can classify text into positive, negative, or neutral categories efficiently. However, it may not perform well on complex sentences, sarcasm, or context-dependent meanings. Overall, it is a useful tool for basic sentiment analysis tasks.
