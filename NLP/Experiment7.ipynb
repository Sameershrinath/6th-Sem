# Experiment 7: TF-IDF Vectorization

## Aim
To implement TF-IDF vectorization using Python in Jupyter Notebook.

## Theory
TF-IDF (Term Frequency–Inverse Document Frequency) is a numerical technique used in Natural Language Processing to evaluate the importance of a word in a document relative to a collection of documents. Term Frequency (TF) measures how often a word appears in a document, while Inverse Document Frequency (IDF) reduces the importance of words that appear frequently across all documents. The TF-IDF value increases with the number of times a word appears in a document but decreases if the word is common in many documents. This helps in identifying important words while ignoring less meaningful ones. TF-IDF is widely used in text mining, information retrieval, and search engines.

## Algorithm / Steps

1. Collect the dataset (documents).
2. Convert text into lowercase.
3. Tokenize the text into words.
4. Calculate Term Frequency (TF) for each word in a document.
5. Calculate Inverse Document Frequency (IDF) for all words.
6. Multiply TF and IDF to get TF-IDF score.
7. Represent each document as a TF-IDF vector.
8. Display the TF-IDF matrix.

```python
# Import required library
from sklearn.feature_extraction.text import TfidfVectorizer

# Dataset (same as previous experiment)
documents = [
    "NLP is very interesting",
    "Machine learning is powerful",
    "NLP and machine learning are related",
    "Deep learning is part of AI",
    "AI is transforming the world",
    "Students are learning NLP",
    "Python is used for machine learning",
    "NLP applications are growing",
    "Data science uses machine learning",
    "AI and NLP are closely related"
]

# Initialize TF-IDF Vectorizer
vectorizer = TfidfVectorizer()

# Fit and transform the dataset
X = vectorizer.fit_transform(documents)

# Get feature names (vocabulary)
features = vectorizer.get_feature_names_out()

# Convert to array
tfidf_matrix = X.toarray()

# Print results
print("Vocabulary:\n", features)
print("\nTF-IDF Matrix:\n", tfidf_matrix)
```

## Output

- Vocabulary: List of all unique words in the dataset.
- TF-IDF Matrix: Numerical representation of documents based on importance of words.
- Each row represents a document.
- Each column represents a word.
- Values are TF-IDF scores (not counts like BoW).
- Higher values indicate more important words in a document.

## Conclusion

TF-IDF vectorization provides a better representation of text data compared to the Bag of Words model by considering the importance of words across all documents. It reduces the weight of commonly occurring words and highlights significant terms. This makes it more effective for tasks like text classification and information retrieval. However, it still does not capture word order or semantic meaning. Overall, TF-IDF is a powerful and widely used technique in NLP.
